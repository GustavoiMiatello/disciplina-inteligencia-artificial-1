```python
import numpy as np

array1 = np.random.randint(1, 10, 20)
array2 = np.random.randint(1, 10, 20)

produto = np.dot(array1, array2)

tamanho1 = np.linalg.norm(array1)
tamanho2 = np.linalg.norm(array2)

coseno = produto / (tamanho1 * tamanho2)

angulo = np.arccos(coseno)

print("Array 1:", array1)
print("Array 2:", array2)
print("Ângulo em radianos:", angulo)
print("Ângulo em graus:", np.degrees(angulo))
```

```python
from scipy.optimize import fsolve

def funcao(x):
    return x**4 - 3*x**3 + 2

raiz1 = fsolve(funcao, 1)[0]
raiz2 = fsolve(funcao, 3)[0]

print("Raiz 1:", raiz1)
print("Raiz 2:", raiz2)
```

```python
from scipy.optimize import minimize_scalar

def f(x):
    return -( -x**2 + 4*x + 1 )

resultado = minimize_scalar(f)

x = resultado.x
valor = -resultado.fun

print("Valor de x:", x)
print("Valor máximo:", valor)
```
